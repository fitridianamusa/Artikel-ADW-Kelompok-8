# Analisis Time Series - Peramalan Harga Telur Semarang menggunakan ARIMA

**Tahapan:**
1. Import & eksplorasi data (EDA)
2. Split data train-test , sebelum semua tahap identifikasi model
3. Uji stasioneritas (ADF Test) → dilakukan pada data train
5. Differencing (jika data belum stasioner),dilakukan pada data train
6. Identifikasi order (p, d, q) via ACF/PACF, dilakukan pada data train.
7. Estimasi/fit model ARIMA pada data train.
8. Evaluasi model pada data test (MAE, RMSE, MAPE).
9. Diagnostik residual model (uji white noise/Ljung-Box, uji normalitas, dan pemeriksaan visual residual).
10. Forecast/peramalan periode ke depan (refit pakai seluruh data).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.seasonal import seasonal_decompose, STL
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.stats.diagnostic import acorr_ljungbox
from sklearn.metrics import mean_absolute_error, mean_squared_error

%matplotlib inline
```

## 1. Import & Eksplorasi Data (EDA)

```python
import pandas as pd

# URL file Excel dari GitHub
url = "https://raw.githubusercontent.com/rahminurulainunfitrah/Praktikum-Analisis-Deret-Waktu/main/harga_telur_semarang.xlsx"

# Membaca file Excel
df = pd.read_excel(url)

# Membersihkan nama kolom
df.columns = df.columns.str.strip().str.lower()

# Mengubah kolom periode menjadi format tanggal
df["periode"] = pd.to_datetime(df["periode"])

# Mengurutkan data berdasarkan periode
df = df.sort_values("periode").reset_index(drop=True)

# Menjadikan periode sebagai index
df = df.set_index("periode")

# Mengubah frekuensi data menjadi harian
df = df.asfreq("D")

# Interpolasi data yang hilang
df["semarang"] = df["semarang"].interpolate()

# Menampilkan jumlah observasi
print("Jumlah observasi:", len(df))

# Menampilkan data
print("\nData:")
display(df)

# Menampilkan statistik deskriptif
print("\nStatistik deskriptif:")
display(df.describe())
```

```python
plt.figure(figsize=(12, 4))
plt.plot(df.index, df["semarang"])
plt.title("Harga Telur Semarang - Data Harian")
plt.xlabel("Tanggal")
plt.ylabel("Harga (Rp)")
plt.tight_layout()
plt.show()
```

## 2. Split Data Train-Test

```python
n_test = 61   # 61 hari terakhir dipakai sebagai data uji, sisanya untuk data latih
train = df["semarang"].iloc[:-n_test]
test = df["semarang"].iloc[-n_test:]

print(f"Jumlah data train : {len(train)}")
print(f"Jumlah data test   : {len(test)}")
```

## Uji Stasioneritas Data

### 1. Uji Stasioneritas Data dalam Nilaitengah

```python
# Uji Stasioneritas Data dalam Nilaitengah
# uji unit-root Augmented Dickey Fuller
from statsmodels.tsa.stattools import adfuller

train = df["semarang"].iloc[:-n_test]
adfuller(train)[1]
```

### 2. Cek Stasioneritas dalam Ragam

```python
train = df["semarang"].iloc[:-n_test]
train.min()
```

```python
# Mencari lambda Box-Cox
from scipy.stats import boxcox, boxcox_llf

train = df["semarang"].iloc[:-n_test]
data_boxcox, lambda_fit, ci = boxcox(
    train,
    alpha=0.05
)

print('Nilai lambda:', lambda_fit)
print('Interval kepercayaan lambda:', ci)
```

```python
# Plot Box-Cox likelihood
# Rentang lambda dibuat mencakup lambda optimal dan CI
train = df["semarang"].iloc[:-n_test]
lambdas = np.linspace(ci[0] - 0.5, ci[1] + 0.5, 300)

llf = [
    boxcox_llf(lmbda,train)
    for lmbda in lambdas
]

plt.figure(figsize=(10, 6))

plt.plot(lambdas, llf)

plt.axvline(
    lambda_fit,
    linestyle='--',
    label=f'Optimal Lambda: {lambda_fit:.2f}'
)

plt.axvspan(
    ci[0],
    ci[1],
    alpha=0.3,
    label=f'95% Confidence Interval: ({ci[0]:.2f}, {ci[1]:.2f})'
)

plt.xlabel('Lambda')
plt.ylabel('Log-Likelihood Function')
plt.title('Box-Cox Transformation Plot')
plt.legend()
plt.grid(True)

plt.show()
```

### 3. Proses Differencing

```python
from statsmodels.tsa.stattools import adfuller

# Data train
train = df["semarang"].iloc[:-n_test]

# Differencing orde 1
train_diff = train.diff().dropna()

# Uji ADF setelah differencing
adf_result = adfuller(train_diff)

print("Hasil Uji ADF Setelah Differencing Orde 1")
print("-----------------------------------------")
print(f"ADF Statistic : {adf_result[0]:.4f}")
print(f"p-value       : {adf_result[1]:.6g}")
```

```python
# Plot Data setelah Differencing Orde 1
plt.figure(figsize=(14, 6))
plt.plot(train_diff)

plt.title('Harga Telur Semarang Setelah Differencing Orde 1')
plt.xlabel('Periode')
plt.ylabel('Differencing Harga')

plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

def acf_pacf_fig(data, both=True, lag=30):
    if both:
        fig, axes = plt.subplots(1, 2, figsize=(14, 5))
        # ACF
        plot_acf(data, lags=lag, ax=axes[0])
        axes[0].set_title("ACF")
        axes[0].set_xlabel("Lag")
        axes[0].set_ylabel("Autocorrelation")

        # PACF
        plot_pacf(data, lags=lag, ax=axes[1], method="ywm")
        axes[1].set_title("PACF")
        axes[1].set_xlabel("Lag")
        axes[1].set_ylabel("Partial Autocorrelation")

        plt.tight_layout()
        plt.show()
```

```python
# Plot ACF dan PACF
acf_pacf_fig(
    train_diff,
    both=True,
    lag=30
)
```

Karena ACF dan PACF sama-sama menunjukkan cutt off di lag 2, kandidat model yang wajar untuk diuji adalah ARIMA(2,1,2). Tetapi karena ada beberapa spike signifikan kembali pada lag yang lebih tinggi, sebaiknya jangan langsung menyimpulkan \(p=q=2\); gunakan beberapa kandidat seperti ARIMA(2,1,0), ARIMA(0,1,2), dan ARIMA(2,1,2) kemudian bandingkan berdasarkan AIC/BIC dan diagnostik residual.

## 4. Model ARIMA

```python
from statsmodels.tsa.arima.model import ARIMA

models = {
    "ARIMA(2,1,0)": ARIMA(train, order=(2,1,0)).fit(),
    "ARIMA(0,1,2)": ARIMA(train, order=(0,1,2)).fit(),
    "ARIMA(2,1,2)": ARIMA(train, order=(2,1,2)).fit(),
    "ARIMA(1,1,2)": ARIMA(train, order=(1,1,2)).fit(),
    "ARIMA(2,1,1)": ARIMA(train, order=(2,1,1)).fit(),
    "ARIMA(1,1,1)": ARIMA(train, order=(1,1,1)).fit()
}
```

```python
for name, result in models.items():
    print(
        name,
        " | AIC =", result.aic,
        " | BIC =", result.bic
    )
```

```python
print(models["ARIMA(2,1,0)"].summary())
print(models["ARIMA(0,1,2)"].summary())
print(models["ARIMA(1,1,2)"].summary())
print(models["ARIMA(2,1,2)"].summary())
print(models["ARIMA(1,1,1)"].summary())
print(models["ARIMA(2,1,1)"].summary())
```

```python
n_test = 61  # 61 hari terakhir dipakai sebagai data uji

train = df["semarang"].iloc[:-n_test]
test = df["semarang"].iloc[-n_test:]

from sklearn.metrics import mean_absolute_percentage_error
import pandas as pd

# forecast data testing
forecast = {}

for name, model in models.items():
    forecast[name] = model.forecast(steps=len(test))

# MAPE
hasil_mape = []

for name, pred in forecast.items():
    mape = mean_absolute_percentage_error(test, pred) * 100

    hasil_mape.append({
        "Model": name,
        "MAPE (%)": mape
    })

# Tabel hasil
hasil_mape = pd.DataFrame(hasil_mape)

print(hasil_mape)
```

## 5. Diagnostik Residual Model (Uji Asumsi & White Noise)

Sebelum model digunakan untuk peramalan, model terbaik hasil perbandingan AIC/BIC dan MAPE pada tahap sebelumnya, yaitu **ARIMA(0,1,2)**, perlu diuji kelayakannya melalui diagnostik residual. Asumsi yang harus dipenuhi oleh residual model ARIMA yang baik adalah:

1. **White noise** — residual tidak memiliki autokorelasi (diuji dengan Ljung-Box test).
2. **Berdistribusi normal** — residual menyebar normal dengan rata-rata nol (diuji dengan Jarque-Bera/Shapiro-Wilk dan QQ-Plot).
3. **Tidak ada pola sistematis** — diperiksa secara visual melalui plot residual dan correlogram.


```python
# Model terbaik berdasarkan AIC/BIC dan MAPE terendah pada tahap sebelumnya
best_model_name = "ARIMA(0,1,2)"
best_model = models[best_model_name]
residual = best_model.resid

print(f"Model terpilih untuk diagnostik residual: {best_model_name}")
print(f"Jumlah residual : {len(residual)}")
print(f"Rata-rata residual : {residual.mean():.4f}")
print(f"Std. residual       : {residual.std():.4f}")

plt.figure(figsize=(14, 4))
plt.plot(residual)
plt.axhline(0, color="red", linestyle="--")
plt.title(f"Plot Residual Model {best_model_name}")
plt.xlabel("Periode")
plt.ylabel("Residual")
plt.tight_layout()
plt.show()
```

### 5.1 Uji White Noise (Uji Autokorelasi Residual — Ljung-Box)

**Hipotesis:**
- H0 : residual tidak memiliki autokorelasi (residual bersifat *white noise*)
- H1 : residual memiliki autokorelasi pada minimal satu lag

**Kriteria keputusan:** jika p-value > alpha (0.05), maka H0 gagal ditolak, artinya residual white noise.


```python
from statsmodels.stats.diagnostic import acorr_ljungbox

lb_test = acorr_ljungbox(residual, lags=[10, 20, 30], return_df=True)
print("Hasil Uji Ljung-Box (White Noise Test)")
print(lb_test)

alpha = 0.05
if (lb_test["lb_pvalue"] > alpha).all():
    print(f"\nSeluruh p-value > {alpha} -> H0 gagal ditolak -> residual bersifat WHITE NOISE.")
else:
    print(f"\nTerdapat p-value <= {alpha} -> H0 ditolak -> residual TIDAK white noise (masih ada autokorelasi).")
```

```python
# Pemeriksaan visual autokorelasi residual (ACF & PACF)
acf_pacf_fig(residual, both=True, lag=30)
```

### 5.2 Uji Normalitas Residual

**Hipotesis:**
- H0 : residual berdistribusi normal
- H1 : residual tidak berdistribusi normal

**Kriteria keputusan:** jika p-value > alpha (0.05), maka H0 gagal ditolak, artinya residual berdistribusi normal. Uji dilakukan dengan dua metode (Jarque-Bera dan Shapiro-Wilk) serta didukung histogram dan QQ-Plot.


```python
from statsmodels.stats.stattools import jarque_bera
from scipy.stats import shapiro

# Uji Jarque-Bera
jb_stat, jb_pvalue, skew, kurtosis = jarque_bera(residual)
print("Uji Jarque-Bera")
print(f"Statistik JB : {jb_stat:.4f}")
print(f"p-value      : {jb_pvalue:.6g}")
print(f"Skewness     : {skew:.4f}")
print(f"Kurtosis     : {kurtosis:.4f}")

# Uji Shapiro-Wilk
sw_stat, sw_pvalue = shapiro(residual)
print("\nUji Shapiro-Wilk")
print(f"Statistik W  : {sw_stat:.4f}")
print(f"p-value      : {sw_pvalue:.6g}")

alpha = 0.05
if jb_pvalue > alpha and sw_pvalue > alpha:
    print(f"\np-value > {alpha} pada kedua uji -> H0 gagal ditolak -> residual berdistribusi NORMAL.")
else:
    print(f"\np-value <= {alpha} -> H0 ditolak -> residual TIDAK berdistribusi normal.")
```

```python
from statsmodels.stats.diagnostic import lilliefors

# Uji Kolmogorov-Smirnov untuk normalitas.
# Catatan: KS test standar mengasumsikan parameter (mean, std) distribusi acuan SUDAH diketahui,
# bukan diestimasi dari data. Karena mean & std di sini diestimasi dari residual itu sendiri,
# digunakan koreksi Lilliefors agar p-value valid.
ks_stat, ks_pvalue = lilliefors(residual, dist="norm")

print("Uji Kolmogorov-Smirnov (dengan koreksi Lilliefors)")
print(f"Statistik D  : {ks_stat:.4f}")
print(f"p-value      : {ks_pvalue:.6g}")

alpha = 0.05
if ks_pvalue > alpha:
    print(f"\np-value > {alpha} -> H0 gagal ditolak -> residual berdistribusi NORMAL (uji KS).")
else:
    print(f"\np-value <= {alpha} -> H0 ditolak -> residual TIDAK berdistribusi normal (uji KS).")
```

```python
from statsmodels.graphics.gofplots import qqplot

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(residual, bins=30, edgecolor="black")
axes[0].set_title("Histogram Residual")
axes[0].set_xlabel("Residual")
axes[0].set_ylabel("Frekuensi")

qqplot(residual, line="s", ax=axes[1])
axes[1].set_title("QQ-Plot Residual")

plt.tight_layout()
plt.show()
```

> **Catatan:** Jika hasil uji normalitas menolak H0 (p-value sangat kecil), periksa histogram dan QQ-Plot di atas — biasanya hal ini disebabkan oleh satu atau beberapa residual outlier (nilai ekstrem), bukan karena bentuk sebaran yang jauh dari normal secara keseluruhan. Perlu dicatat bahwa estimasi parameter ARIMA dengan Maximum Likelihood tetap valid secara asimtotik meski normalitas sedikit menyimpang, namun interval kepercayaan/forecast interval perlu diinterpretasikan dengan hati-hati apabila asumsi ini tidak terpenuhi dengan baik.

### 5.4 Ringkasan Uji Asumsi Residual

| Uji | Tujuan | Kriteria | Kesimpulan |

| Jarque-Bera, Shapiro-Wilk & Kolmogorov-Smirnov (Lilliefors) | Normalitas residual | p-value > 0.05 | Lihat output Bagian 5.2 |

Jika salah satu asumsi tidak terpenuhi (misalnya normalitas), model masih dapat digunakan untuk menghasilkan titik ramalan (point forecast), namun interval kepercayaan ramalan perlu diinterpretasikan dengan lebih hati-hati, atau dapat dipertimbangkan penanganan outlier/transformasi data sebagai tindak lanjut.
